# Basic Data Exploration

### Load the data

This Python code imports the pandas library, reads a CSV file into a DataFrame and displays the DataFrame's contents.

```
import pandas as pd

# Read the CSV file into a DataFrame
df = pd.read_csv('customer_churn_data.csv')

# Display the DataFrame
print(df)
```

### Checking Dimensions -  typically refers to examining the shape or structure of a dataset

```
# Check the dimensions of the DataFrame
print(df.shape)
```

```
# View the first few rows of the DataFrame
print(df.head())
```

```
# View the first few rows of the DataFrame
print(df.head(10))
```

```
# View the last few rows of the DataFrame
print(df.tail())
```

```
# View the last few rows of the DataFrame
print(df.tail(10))
```

### To generate summary statistics for your DataFrame you can use the describe() method.

```
# Generate summary statistics for the DataFrame
print(df.describe())
```

### To display the data types of each column in your DataFrame you can use the dtypes attribute.

```
# Display the data types of each column in the DataFrame
print(df.dtypes)
```

```
# Display the basic information of each column in the DataFrame
print(df.info())
```

```
summation = df[['Age', 'Tenure_Months', 'Total_Charges']].sum()
print(summation)
```

```
average = df[['Age', 'Tenure_Months', 'Total_Charges']].mean()
print(average)
```

```
minimum = df[['Age', 'Tenure_Months', 'Total_Charges']].min()
print(minimum)
```

```
maximum = df[['Age', 'Tenure_Months', 'Total_Charges']].max()
print(maximum)
```

### To count the number of missing values in each column of your DataFrame you can use the isnull() method followed by the sum() method.

```
# Count the number of missing values in each column of the DataFrame
missing_values_count = df.isnull().sum()

print(missing_values_count)
```

### To get the count of unique values for a categorical column in your DataFrame you can use the value_counts()

```
# Replace 'df' with the name of your DataFrame and 'categorical_column' with any column name from your dataset
unique_values = df['Payment_Method'].value_counts()

print("Unique Values (for categorical column):")
print(unique_values)
```

### To create a histogram of the data distribution for a numeric column in your DataFrame using matplotlib

```
import matplotlib.pyplot as plt

# Histogram for a categorical feature
df['Payment_Method'].hist()

plt.show()
```

### Correlation: df.corr() -  calculates the correlation coefficients between all pairs of numeric columns in the DataFrame and prints the correlation matrix. Each cell in the correlation matrix represents the correlation coefficient between the corresponding pair of columns. The correlation coefficient ranges from -1 to 1, where:

- 1 indicates a perfect positive correlation,
- -1 indicates a perfect negative correlation, and
- 0 indicates no correlation.

```
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate the correlation matrix
correlations = df[['Age', 'Tenure_Months', 'Monthly_Charges', 'Total_Charges']].corr()
print(correlations)

# Heatmap of the correlation
sns.heatmap(correlations, annot=True, cmap='coolwarm', linewidths=0.5)
```

```
df.boxplot(column='Age')
plt.show()
```

```
df.boxplot(column=['Age', 'Tenure_Months'])
plt.show()
```

### Deletion

### Remove columns with many missing values

```
df_weather = pd.read_csv('ny_weather.csv')

null_values = df_weather.isnull().sum()
print(null_values)
```

```
# Remove columns with many missing values
df_weather.drop('Temperature', axis=1, inplace=True)
print(df_weather.info())
```

### Imputation

### Forward or backward Fill missing values with the mean


### Forward Fill - In the forward fill method, missing values are filled with the last observed non-null value in each column.

```
df_weather = pd.read_csv('ny_weather.csv')
print(df_weather)
```

```
df_weather.ffill(inplace=True) # Inplace=True means, it will modify the original dataset
print(df_weather.isnull().sum())
print(df_weather)
```

### Backward Fill - In the backward fill method, missing values are filled with the next observed non-null value in each column.

```
df_weather = pd.read_csv('ny_weather.csv')
print(df_weather.isnull().sum())
```

```
df_weather.bfill(inplace=True) # Inplace=True means, it will modify the original dataset
print(df_weather.isnull().sum())
print(df_weather)
```

### Indicator Variables


### In the context of missing data, indicator variables can be used to flag whether a certain value is missing or not. This can be helpful for analysis and modeling, as missing data often needs to be handled or accounted for in some way.

### This code creates a new column named 'column_name_missing' in the DataFrame. The purpose of this column is to indicate whether the corresponding value in the column 'column_name' is missing or not.

```
df_weather = pd.read_csv('ny_weather.csv')

# Indicator Variables
df_weather['Temperature_missing'] = df_weather['Temperature'].isnull().astype(int)

print(df_weather)
```

# Creating Basic Plot


### Line plot

A line plot is a type of graph that displays data points connected by straight lines. It is commonly used to visualize the relationship between two variables over a continuous interval.

```
df_sales = pd.read_csv('sales.csv')
df_sales.head()
```

```
import matplotlib.pyplot as plt

# Create a line plot
plt.plot(df_sales['Sales_Revenue'], df_sales['Units_Sold'], marker='o', linestyle='-', color='b', label='Line Plot')

# Adding labels and a title
plt.xlabel('Sales_Revenue')  # Label for the X-axis
plt.ylabel('Units_Sold')  # Label for the Y-axis
plt.title('Line Plot Example')  # Title of the plot

# Adding a legend
plt.legend()  # Show the legend with label specified in the plot

# Show the plot
plt.show()  # Display the plot

```

### Scatter plot

A scatter plot is a type of data visualization that displays the relationship between two variables by representing data points as individual dots on a two-dimensional plane

```
import matplotlib.pyplot as plt

# Create a scatter plot
plt.scatter(df_sales['Sales_Revenue'], df_sales['Units_Sold'], marker='o', linestyle='-', color='b', label='Line Plot')

# Adding labels and a title
plt.xlabel('Sales_Revenue')  # Label for the X-axis
plt.ylabel('Units_Sold')  # Label for the Y-axis
plt.title('Line Plot Example')  # Title of the plot

plt.legend()  # Show legend

# Show the plot
plt.show()

```

### Bar Chart

A bar chart is a type of graph that presents categorical data with rectangular bars. The length or height of each bar is proportional to the value it represents.

```
df = pd.read_csv('customer_churn_data.csv')
df.head()
```

```
# Extracting Value Counts from the Dataset
data_dict = df['Payment_Method'].value_counts().to_dict()
print(data_dict)
```

```
import matplotlib.pyplot as plt

Payment_Method = data_dict.keys()
counts = data_dict.values()

# Create a bar chart
plt.bar(Payment_Method, counts, color='g', alpha=0.7, label='Bar Chart')

# Adding labels and a legend
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Bar Chart Example')
plt.legend()

# Show the plot
plt.show()

```

#  Saving Plots


 - Import Matplotlib.
 - Create your plot using plt.plot(), plt.scatter(), or any other plot function.
 - Customize the plot as needed (e.g., add labels and a title).
 - Use plt.savefig() to save the plot as an image file. You provide the file name with the desired file extension (e.g., 'my_plot.png', 'my_plot.pdf').


### Save the plot as an image file (e.g., PNG)

```
import matplotlib.pyplot as plt

# Sample data
x = [1, 2, 3, 4, 5]
y = [10, 14, 8, 20, 12]

# Create a scatter plot
plt.scatter(x, y, marker='o')

# Add text annotation to the plot
plt.text(2, 25, 'Annotation Text', fontsize=12, color='red')

# Save the plot as an image file (PNG)
plt.savefig('my_plot.png')

# Display the plot
plt.show()

```

# Seaborn

### Seaborn is a data visualization library.
 - Built on top of Matplotlib.
 - Designed for creating beautiful and informative statistical plots.


### Creating Statistical Plots


### Histogram

A histogram is a graphical representation of the distribution of numerical data. It consists of a series of adjacent rectangles, or bins, each representing a specific interval or range of values, and the height of each rectangle corresponds to the frequency, or count, of data points falling within that interval.

```

import seaborn as sns
import matplotlib.pyplot as plt

# Sample data
data = sns.load_dataset("iris")

# Create a histogram
sns.histplot(data=data, x="sepal_length", kde=True)

# Adding labels and a title
plt.xlabel("Sepal Length")
plt.ylabel("Frequency")
plt.title("Histogram of Sepal Length")

# Show the plot
plt.show()
```

### Box Plot

A box plot is a method for graphically depicting groups of numerical data through their quartiles, providing a visual summary of the distribution, outliers, and variability.

```
import seaborn as sns
import matplotlib.pyplot as plt

# Sample data
data = sns.load_dataset("tips")

# Create a box plot
sns.boxplot(data=data, x="day", y="total_bill", palette="Set2")

# Adding labels and a title
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.title("Box Plot of Total Bill by Day")

# Show the plot
plt.show()
```

### Violin Plot

A violin plot is a method of plotting numeric data and can be considered a combination of a box plot and a kernel density plot. It displays the distribution of the data across different levels of a categorical variable.

```
import seaborn as sns
import matplotlib.pyplot as plt

# Sample data
data = sns.load_dataset("tips")

# Create a violin plot
sns.violinplot(data=data, x="day", y="total_bill", palette="pastel")

# Adding labels and a title
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.title("Violin Plot of Total Bill by Day")

# Show the plot
plt.show()
```

### Pair plot

A pair plot, also known as a scatterplot matrix, is a grid of scatterplots that displays the pairwise relationships between variables in a dataset. Each scatterplot in the grid represents the relationship between two variables, allowing for visual examination of potential correlations and patterns across multiple variables simultaneously. It's a useful tool for exploring the relationships between multiple variables in a dataset.







```
import seaborn as sns
import matplotlib.pyplot as plt

# Load a sample dataset
data = sns.load_dataset("iris")

# Create a pair plot
sns.pairplot(data, hue="species", markers=["o", "s", "D"])

# Show the plot
plt.show()
```

### Heat Map

A heatmap is a graphical representation of data where values in a matrix are represented as colors. It's particularly useful for visualizing relationships between two variables in a dataset. In a heatmap, each cell in the matrix is assigned a color based on its value. The colors typically range from dark (representing low values) to light (representing high values), allowing patterns and trends to be easily identified.

```
import seaborn as sns
import matplotlib.pyplot as plt

# Load the "tips" dataset
data = sns.load_dataset("tips")

# Select relevant numeric columns
numeric_data = data[['total_bill', 'tip', 'size']]

# Calculate the correlation matrix
correlation_matrix = numeric_data.corr()

# Create a heat map
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm")

# Show the plot
plt.show()
```
